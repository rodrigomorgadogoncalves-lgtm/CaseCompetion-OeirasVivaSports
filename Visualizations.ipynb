```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

plt.rcParams.update({
    "figure.facecolor": "#ffffff",
    "axes.facecolor": "#ffffff",
    "axes.edgecolor": "#d0d0d0",
    "axes.labelcolor": "#222222",
    "text.color": "#222222",
    "xtick.color": "#444444",
    "ytick.color": "#444444",
    "grid.color": "#d9d9d9",
    "grid.alpha": 0.8,
    "font.family": "DejaVu Sans",
    "font.size": 11,
    "figure.dpi": 140,
})

PALETTE = ["#4f46e5", "#06b6d4", "#ec4899", "#eab308"]

# 1. Evolução de Utentes por Atividade
categories = ["Pavilhões\nJogos Oficiais", "Pavilhões\nTreinos",
              "Piscinas\nutentes", "Viva\nFérias"]
data_2022 = [900, 6500, 2600, 1000]
data_2023 = [1092, 7228, 3596, 1188]
data_2024 = [2295, 7812, 3870, 1328]

fig, ax = plt.subplots(figsize=(12, 6.5))
x = np.arange(len(categories))
w = 0.24

for i, (vals, year, col) in enumerate(
    zip([data_2022, data_2023, data_2024], ["2022", "2023", "2024"], PALETTE[:3])
):
    bars = ax.bar(
        x + (i - 1) * w, vals, w,
        label=year,
        color=col,
        edgecolor=col,
        linewidth=0.5,
        zorder=3,
        alpha=0.92
    )
    for bar in bars:
        h = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            h + 80,
            f"{h:,.0f}",
            ha="center",
            va="bottom",
            fontsize=8.5,
            fontweight="bold",
            color=col
        )

ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=10.5)
ax.set_ylabel("Utentes", fontsize=12, fontweight="bold")
ax.set_title("Evolução de Utentes por Atividade  ·  2022 → 2024",
             fontsize=16, fontweight="bold", pad=18)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f"{v:,.0f}"))
ax.legend(frameon=False, fontsize=10, loc="upper left")
ax.grid(axis="y", linestyle="--", linewidth=0.4, zorder=0)
ax.set_axisbelow(True)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig("01_utentes_atividade.png", bbox_inches="tight", facecolor="white")
plt.show()

# 2. Facebook Viva  ·  Aumento percentual 2023 → 2024
fb_metrics = ["Alcance", "Visitas", "Interação\nconteúdos", "Seguidores"]
fb_2023 = np.array([79802, 18800, 10800, 10200], dtype=float)
fb_2024 = np.array([167000, 34000, 13400, 10800], dtype=float)

fb_growth = (fb_2024 - fb_2023) / fb_2023 * 100

fig, ax = plt.subplots(figsize=(10, 6))
x = np.arange(len(fb_metrics))

bars = ax.bar(
    x, fb_growth,
    color=["#4f46e5", "#06b6d4", "#ec4899", "#eab308"],
    edgecolor=["#4f46e5", "#06b6d4", "#ec4899", "#eab308"],
    linewidth=0.6,
    alpha=0.9,
    zorder=3
)

ax.axhline(0, color="#999999", linewidth=0.8)

for bar, val in zip(bars, fb_growth):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        val + 4,
        f"{val:.1f}%",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
        color="#222222"
    )

ax.set_xticks(x)
ax.set_xticklabels(fb_metrics, fontsize=10)
ax.set_ylabel("Aumento percentual (%)", fontsize=11, fontweight="bold")
ax.set_title("Facebook Viva  ·  Crescimento 2023 → 2024",
             fontsize=15, fontweight="bold", pad=16)
ax.set_ylim(0, max(fb_growth) + 20)
ax.yaxis.set_major_formatter(mticker.PercentFormatter())
ax.grid(axis="y", linestyle="--", linewidth=0.3, zorder=0)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig("02_facebook_viva_crescimento.png", bbox_inches="tight", facecolor="white")
plt.show()

# 3. Instagram Viva  ·  Aumento percentual 2023 → 2024
ig_metrics = ["Alcance", "Visitas", "Seguidores"]
ig_2023 = np.array([8796, 5400, 2700], dtype=float)
ig_2024 = np.array([33000, 9000, 3800], dtype=float)

ig_growth = (ig_2024 - ig_2023) / ig_2023 * 100

fig, ax = plt.subplots(figsize=(10, 6))
x = np.arange(len(ig_metrics))

bars = ax.bar(
    x, ig_growth,
    color=["#ec4899", "#eab308", "#06b6d4"],
    edgecolor=["#ec4899", "#eab308", "#06b6d4"],
    linewidth=0.6,
    alpha=0.9,
    zorder=3
)

ax.axhline(0, color="#999999", linewidth=0.8)

for bar, val in zip(bars, ig_growth):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        val + 5,
        f"{val:.1f}%",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
        color="#222222"
    )

ax.set_xticks(x)
ax.set_xticklabels(ig_metrics, fontsize=10)
ax.set_ylabel("Aumento percentual (%)", fontsize=11, fontweight="bold")
ax.set_title("Instagram Viva  ·  Crescimento 2023 → 2024",
             fontsize=15, fontweight="bold", pad=16)
ax.set_ylim(0, max(ig_growth) + 30)
ax.yaxis.set_major_formatter(mticker.PercentFormatter())
ax.grid(axis="y", linestyle="--", linewidth=0.3, zorder=0)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig("03_instagram_viva_crescimento.png", bbox_inches="tight", facecolor="white")
plt.show()
```
